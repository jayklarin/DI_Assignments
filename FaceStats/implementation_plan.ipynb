# 📘 Implementation Plan — Version 1 (Original Face Stats Idea)

## 1. Core Concept
Face Stats was originally conceived as an app that:
- Generates or retrieves a set of portrait images.
- Rates each face on attractiveness metrics such as symmetry and proportions.
- Produces a 1–10 attractiveness rating.
- Displays distributions, bell curves, mean/median/std, and other statistical visualizations.
- Enables compositing / averaging a selection of faces into a “mean face” based on chosen filters.

## 2. Dataset Strategy (Early Vision)
- Source portraits from generative services or pre-existing synthetic datasets.
- Keep dataset size manageable (100–1000 images).
- Store only raw images and basic metadata.
- No full preprocessing pipeline yet — rely on already-aligned images when possible.

## 3. Attractiveness Metrics (Initial Set)
- Symmetry
- Golden Ratio proportionality
- Skin clarity
- Eye spacing
- Facial proportions (upper/mid/lower thirds)
- Optional: Age cues, gender cues

Scoring model: either manual heuristics or a pretrained embedding-based beauty predictor.

## 4. Statistical Analysis (Original)
- Compute distributions of attractiveness scores.
- Plot histograms, KDEs, bell curves.
- Compute summary statistics (mean, median, mode, variance, std).

## 5. Compositing (Early Idea)
- User selects N faces (or filters like 1–3 rating, 8–10 rating, etc.).
- Compute the pixel-wise or embedding-wise average to create the composite.
- Show mean face vs. representative examples.
- Compare composite scores to population distribution.

## 6. UI / Workflow (Original)
- A simple Streamlit or notebook interface displaying:
  - Dataset preview
  - Ratings table
  - Distribution graphs
  - Composite face generation panel
  - Sliders for filtering by attractiveness, gender, age, etc.

## 7. MVP Deliverable
- A functioning prototype that:
  - Loads a folder of portraits.
  - Rates each portrait on key attractiveness metrics.
  - Outputs distributions.
  - Performs a basic composite of N portraits.
  - Shows comparative statistics.

This Version 1 reflects the earliest vision before deeper structuring, preprocessing, embeddings, naming conventions, or multi-tab workflows were introduced.

## 3. Preprocessing Pipeline (Final Spec)

### 3.1 Steps
- Load all images from `dataset_raw/`.
- Detect faces with MediaPipe or similar.
- Extract landmarks.
- Compute head pose (yaw/pitch/roll).
- Align faces to canonical orientation.
- Crop + resize to fixed resolution (e.g., 512×512).
- Filter low-quality images (blurry, occluded, extreme angles).
- Save clean output to `dataset_preprocessed/`.

### 3.2 Metadata collected
- Age estimate  
- Gender estimate  
- Head pose  
- Confidence scores  
- Landmark-based ratios  

This metadata enables advanced filtering during compositing and analysis.

---

## 4. Attractiveness Metrics (Expanded & Updated)
- Symmetry (landmark deviation across vertical axis)
- Golden ratio conformity
- Skin evenness (local variance filters)
- Smoothness vs. texture irregularity
- Eye spacing proportionality
- Facial thirds ratios
- Embedding-based beauty scores (CLIP / deep features)
- Distribution-normalized attractiveness score: **1–10 baseline**, enforced globally

---

## 5. Embeddings + Features

## 5.1 Updated Method
- Use a modern facial embedding model (e.g., InsightFace, FaceNet2, or a CLIP-based encoder).
- Save embeddings for each aligned image.
- Store in `embeddings/` and/or a FAISS index.

## 5.2 Analysis Tasks
- PCA of embeddings
- Clustering by attractiveness range
- Outlier detection
- Composite embedding calculation (mean vector)

---

## 6. Compositing Pipeline (Final Spec)

## 6.1 Input Filters
Users can specify:
- `samples = N`
- `score = [min]` or `[min, max]`
- `gender = 'm'/'f'/any`
- `agemin`, `agemax`
- `ethnicity`
- `metadata combinations`

## 6.2 Composite Generation Methods
- Pixel-wise mean face
- Embedding-weighted average face
- PCA-based reconstruction (eigenface-style)
- Weighted composites (e.g., proportional to attractiveness)

## 6.3 Output
- Composite image saved under `/composites/`
- Attractiveness rating of the composite relative to population
- Stats panel comparing composite vs filtered group vs overall dataset

## 7. Statistical Analysis (Updated)

# 📋 **Final Project — Capstone Scope Form (Face Stats)**

## 1. Project Title
**Face Stats — A Statistical & Generative Analysis System for Human-Like Portraits**

---

## 2. Project Description & Justification

Face Stats is a full-stack system that:
- Collects or generates synthetic portrait datasets.
- Preprocesses them (alignment, filtering, metadata extraction).
- Rates each portrait on multiple attractiveness metrics.
- Normalizes all portraits to a 1–10 attractiveness scale.
- Computes embeddings and performs statistical + PCA analysis.
- Generates composite faces based on filters such as attractiveness, age, gender, and ethnicity.

**Justification:**  
Attractiveness analysis is a long-standing topic in cognitive science, psychology, and computer graphics. This project creates a reproducible ML pipeline for studying facial metrics, generating composites, and analyzing trends statistically. It demonstrates mastery of data preprocessing, embeddings, computer vision, ML modeling, visualization, and the entire ML engineering lifecycle.
